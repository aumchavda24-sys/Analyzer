```python
import numpy as np


class DataAnalytics:
    """A NumPy analyzer that demonstrates arrays, OOP, and statistical operations."""

    def __init__(self):
        self._array = None
        self._second_array = None

    @classmethod
    def create_analyzer(cls):
        """Class method for creating a DataAnalytics object."""
        return cls()

    @staticmethod
    def display_title():
        """Static method for displaying the application title."""
        print("\n" + "=" * 39)
        print("          Welcome to the NumPy Analyzer!")
        print("=" * 39)

    def _read_array_values(self, count, label="Enter array elements"):
        while True:
            try:
                values = input(f"{label} ({count} elements separated by space): ").split()
                if len(values) != count:
                    print(f"Please enter exactly {count} values.")
                    continue
                return [float(value) if "." in value else int(value) for value in values]
            except ValueError:
                print("Invalid input. Please enter numeric values.")

    def _create_array(self):
        print("\nSelect the type of array to create:")
        print("1. 1D Array")
        print("2. 2D Array")
        print("3. 3D Array")
        choice = input("Enter your choice: ").strip()

        try:
            if choice == "1":
                size = int(input("Enter the number of elements: "))
                if size <= 0:
                    print("Size must be positive.")
                    return
                self._array = np.array(self._read_array_values(size))
            elif choice == "2":
                rows = int(input("Enter the number of rows: "))
                cols = int(input("Enter the number of columns: "))
                if rows <= 0 or cols <= 0:
                    print("Rows and columns must be positive.")
                    return
                values = self._read_array_values(rows * cols)
                self._array = np.array(values).reshape(rows, cols)
            elif choice == "3":
                depth = int(input("Enter the number of layers: "))
                rows = int(input("Enter the number of rows per layer: "))
                cols = int(input("Enter the number of columns: "))
                if depth <= 0 or rows <= 0 or cols <= 0:
                    print("All dimensions must be positive.")
                    return
                values = self._read_array_values(depth * rows * cols)
                self._array = np.array(values).reshape(depth, rows, cols)
            else:
                print("Invalid option.")
                return

            print("\nArray created successfully:")
            print(self._array)
            print("Shape:", self._array.shape)
            print("Dimensions:", self._array.ndim)
        except ValueError:
            print("Invalid dimensions. Please enter whole numbers.")

    def _require_array(self):
        if self._array is None:
            print("\nPlease create an array first (menu option 1).")
            return False
        return True

    def _index_or_slice(self):
        if not self._require_array():
            return

        print("\nChoose an operation:")
        print("1. Indexing")
        print("2. Slicing")
        print("3. Go Back")
        choice = input("Enter your choice: ").strip()

        try:
            if choice == "1":
                if self._array.ndim == 1:
                    index = int(input(f"Enter index (0 to {len(self._array) - 1}; negative indices allowed): "))
                    print("Element:", self._array[index])
                elif self._array.ndim == 2:
                    row = int(input("Enter row index: "))
                    col = int(input("Enter column index: "))
                    print("Element:", self._array[row, col])
                else:
                    layer = int(input("Enter layer index: "))
                    row = int(input("Enter row index: "))
                    col = int(input("Enter column index: "))
                    print("Element:", self._array[layer, row, col])
            elif choice == "2":
                if self._array.ndim == 1:
                    start = input("Enter slice start (blank for beginning): ").strip()
                    end = input("Enter slice end (blank for end): ").strip()
                    step = input("Enter slice step (blank for 1): ").strip()
                    result = self._array[
                        int(start) if start else None:
                        int(end) if end else None:
                        int(step) if step else None
                    ]
                elif self._array.ndim == 2:
                    row_range = input("Enter row range (start:end), e.g. 0:2: ").strip()
                    col_range = input("Enter column range (start:end), e.g. 1:3: ").strip()
                    r_start, r_end = self._parse_slice(row_range)
                    c_start, c_end = self._parse_slice(col_range)
                    result = self._array[r_start:r_end, c_start:c_end]
                else:
                    print("For a 3D array, enter ranges for layers, rows, and columns.")
                    layer_range = input("Layer range (start:end): ").strip()
                    row_range = input("Row range (start:end): ").strip()
                    col_range = input("Column range (start:end): ").strip()
                    l_start, l_end = self._parse_slice(layer_range)
                    r_start, r_end = self._parse_slice(row_range)
                    c_start, c_end = self._parse_slice(col_range)
                    result = self._array[l_start:l_end, r_start:r_end, c_start:c_end]
                print("\nSliced Array:")
                print(result)
            elif choice == "3":
                return
            else:
                print("Invalid option.")
        except (ValueError, IndexError, TypeError) as error:
            print("Could not perform indexing/slicing:", error)

    @staticmethod
    def _parse_slice(text):
        parts = text.split(":")
        if len(parts) != 2:
            raise ValueError("Use start:end format, for example 0:2.")
        start = int(parts[0]) if parts[0].strip() else None
        end = int(parts[1]) if parts[1].strip() else None
        return start, end

    def _mathematical_operations(self):
        if not self._require_array():
            return
        print("\nChoose a mathematical operation:")
        print("1. Addition")
        print("2. Subtraction")
        print("3. Multiplication (element-wise)")
        print("4. Division")
        print("5. Dot product / matrix multiplication (2D arrays)")
        choice = input("Enter your choice: ").strip()

        try:
            if choice in {"1", "2", "3", "4", "5"}:
                if choice == "5":
                    if self._array.ndim != 2:
                        print("Dot product/matrix multiplication requires a 2D original array.")
                        return
                    rows = int(input("Enter rows for the second matrix: "))
                    cols = int(input("Enter columns for the second matrix: "))
                    if rows <= 0 or cols <= 0:
                        print("Dimensions must be positive.")
                        return
                    values = self._read_array_values(rows * cols, "Enter second matrix elements")
                    other = np.array(values).reshape(rows, cols)
                    if choice == "5":
                        if self._array.shape[1] != other.shape[0]:
                            print("Incompatible shapes for matrix multiplication:", self._array.shape, "and", other.shape)
                            return
                        result = np.dot(self._array, other)
                    print("\nOriginal Array:\n", self._array, sep="")
                    print("\nSecond Array:\n", other, sep="")
                    print("\nResult of Matrix Multiplication:\n", result, sep="")
                    return

                if choice == "4":
                    if np.any(self._array == 0):
                        print("Division by zero is not allowed; the original array contains zero.")
                        return

                count = self._array.size
                values = self._read_array_values(count, "Enter the elements of another array")
                other = np.array(values).reshape(self._array.shape)

                if choice == "1":
                    result = np.add(self._array, other)
                    operation = "Addition"
                elif choice == "2":
                    result = np.subtract(self._array, other)
                    operation = "Subtraction"
                elif choice == "3":
                    result = np.multiply(self._array, other)
                    operation = "Multiplication"
                else:
                    if np.any(other == 0):
                        print("Division by zero is not allowed; the second array contains zero.")
                        return
                    result = np.divide(self._array, other)
                    operation = "Division"

                print("\nOriginal Array:\n", self._array, sep="")
                print("\nSecond Array:\n", other, sep="")
                print(f"\nResult of {operation}:\n{result}")
            else:
                print("Invalid option.")
        except (ValueError, ZeroDivisionError) as error:
            print("Could not perform operation:", error)

    def _combine_or_split(self):
        if not self._require_array():
            return
        print("\nChoose an option:")
        print("1. Combine Arrays")
        print("2. Split Array")
        choice = input("Enter your choice: ").strip()

        try:
            if choice == "1":
                if self._array.ndim == 1:
                    values = self._read_array_values(
                        self._array.size, "Enter the elements of another array to combine"
                    )
                    other = np.array(values)
                    print("\nCombined Array:")
                    print(np.concatenate((self._array, other)))
                else:
                    print("For a 2D/3D array, the second array must have compatible dimensions.")
                    print("Enter values for a second array with the same shape.")
                    values = self._read_array_values(
                        self._array.size, "Enter the elements of another array to combine"
                    )
                    other = np.array(values).reshape(self._array.shape)
                    axis_text = input(
                        f"Enter axis to combine (0 to {self._array.ndim - 1}; 0 = vertical for 2D): "
                    ).strip()
                    axis = int(axis_text)
                    result = np.concatenate((self._array, other), axis=axis)
                    print("\nOriginal Array:\n", self._array, sep="")
                    print("\nSecond Array:\n", other, sep="")
                    print("\nCombined Array:")
                    print(result)
            elif choice == "2":
                axis = 0
                if self._array.ndim > 1:
                    axis = int(input(f"Enter split axis (0 to {self._array.ndim - 1}): "))
                pieces = int(input("Enter number of equal parts to split into: "))
                result = np.array_split(self._array, pieces, axis=axis)
                print("\nSplit Arrays:")
                for index, part in enumerate(result, start=1):
                    print(f"Part {index}:\n{part}")
            else:
                print("Invalid option.")
        except (ValueError, IndexError) as error:
            print("Could not combine or split arrays:", error)

    def _search_sort_filter(self):
        if not self._require_array():
            return
        print("\nChoose an option:")
        print("1. Search a value")
        print("2. Sort the array")
        print("3. Filter values")
        choice = input("Enter your choice: ").strip()

        try:
            if choice == "1":
                value = float(input("Enter value to search for: "))
                positions = np.argwhere(self._array == value)
                if positions.size:
                    print("Value found at index/indices:")
                    print(positions)
                else:
                    print("Value not found.")
            elif choice == "2":
                order = input("Sort order (A = ascending, D = descending): ").strip().upper()
                if order not in {"A", "D"}:
                    print("Invalid sort order.")
                    return
                if self._array.ndim == 1:
                    result = np.sort(self._array)
                    if order == "D":
                        result = result[::-1]
                else:
                    result = np.sort(self._array, axis=-1)
                    if order == "D":
                        result = np.flip(result, axis=-1)
                    print("(Sorting applied along the last axis.)")
                print("\nSorted Array:")
                print(result)
            elif choice == "3":
                operator = input("Condition (>, <, >=, <=, ==, !=): ").strip()
                value = float(input("Filter value: "))
                comparisons = {
                    ">": np.greater,
                    "<": np.less,
                    ">=": np.greater_equal,
                    "<=": np.less_equal,
                    "==": np.equal,
                    "!=": np.not_equal,
                }
                if operator not in comparisons:
                    print("Invalid condition.")
                    return
                result = self._array[comparisons[operator](self._array, value)]
                print("\nFiltered values:")
                print(result)
            else:
                print("Invalid option.")
        except ValueError as error:
            print("Invalid input:", error)

    def _aggregates_and_statistics(self):
        if not self._require_array():
            return
        print("\nChoose an aggregate/statistical operation:")
        print("1. Sum")
        print("2. Mean")
        print("3. Median")
        print("4. Standard Deviation")
        print("5. Variance")
        print("6. Minimum")
        print("7. Maximum")
        print("8. Percentiles")
        print("9. Correlation coefficient with another array")
        choice = input("Enter your choice: ").strip()

        try:
            if choice == "1":
                print("Sum of Array:", np.sum(self._array))
            elif choice == "2":
                print("Mean of Array:", np.mean(self._array))
            elif choice == "3":
                print("Median of Array:", np.median(self._array))
            elif choice == "4":
                print("Standard Deviation of Array:", np.std(self._array))
            elif choice == "5":
                print("Variance of Array:", np.var(self._array))
            elif choice == "6":
                print("Minimum of Array:", np.min(self._array))
            elif choice == "7":
                print("Maximum of Array:", np.max(self._array))
            elif choice == "8":
                percentile = float(input("Enter percentile (0 to 100): "))
                if not 0 <= percentile <= 100:
                    print("Percentile must be between 0 and 100.")
                    return
                print(f"{percentile:g}th Percentile:", np.percentile(self._array, percentile))
            elif choice == "9":
                if self._array.size < 2:
                    print("At least two elements are needed for correlation.")
                    return
                values = self._read_array_values(
                    self._array.size, "Enter elements of another array for correlation"
                )
                other = np.array(values).reshape(self._array.shape)
                correlation = np.corrcoef(self._array.ravel(), other.ravel())[0, 1]
                print("Correlation Coefficient:", correlation)
            else:
                print("Invalid option.")
        except ValueError as error:
            print("Invalid input:", error)

    def run(self):
        self.display_title()
        while True:
            print("\nChoose an option:")
            print("1. Create a NumPy Array")
            print("2. Perform Mathematical Operations")
            print("3. Combine or Split Arrays")
            print("4. Search, Sort, or Filter Arrays")
            print("5. Compute Aggregates and Statistics")
            print("6. Indexing and Slicing")
            print("7. Exit")

            choice = input("Enter your choice: ").strip()
            match choice:
                case "1":
                    self._create_array()
                case "2":
                    self._mathematical_operations()
                case "3":
                    self._combine_or_split()
                case "4":
                    self._search_sort_filter()
                case "5":
                    self._aggregates_and_statistics()
                case "6":
                    self._index_or_slice()
                case "7":
                    print("\nThank you for using the NumPy Analyzer! Goodbye!")
                    break
                case _:
                    print("\nInvalid option. Please select a valid option from the menu.")


# Start the program directly (no main() wrapper).
analyzer = DataAnalytics.create_analyzer()
analyzer.run()
```
